## 🛠️ Helper FunctionsUtility functions for PII detection, summarization, and performance metrics

```python
# Cell 10: Define Comprehensive Helper Functionsimport timeimport hashlibfrom typing import Dict, Listimport psutilimport osdef strip_pii(text: str) -> Dict[str, any]:    """Demonstrate PII stripping (simplified for demo)"""    start_time = time.time()        detected_pii = []    cleaned_text = text        # Simple PII patterns (demo only)    text_lower = text.lower()    pii_patterns = {        "names": ["john", "jane", "patient"],        "dates": ["dob", "date of birth"],        "identifiers": ["ssn", "mrn", "id #"]    }        for category, patterns in pii_patterns.items():        for pattern in patterns:            if pattern in text_lower:                detected_pii.append(f"{category}: {pattern}")        return {        "original_text": text,        "cleaned_text": cleaned_text,        "pii_detected": detected_pii if detected_pii else ["None detected"],        "text_hash": hashlib.sha256(text.encode()).hexdigest()[:16],        "process_time_ms": round((time.time() - start_time) * 1000, 2),        "privacy_compliant": True    }def generate_demo_summary(text: str) -> str:    """Generate demo medical summary"""    text_lower = text.lower()        symptoms = []    conditions = []    medications = []        symptom_kw = ["headache", "pain", "fever", "nausea", "fatigue", "dizzy", "thirst", "cough", "wheezing"]    condition_kw = ["diabetes", "hypertension", "asthma", "migraine", "chronic pain", "back pain"]    med_kw = ["metformin", "lisinopril", "ibuprofen", "albuterol", "fluticasone", "losartan"]        for symptom in symptom_kw:        if symptom in text_lower:            symptoms.append(symptom.capitalize())        for condition in condition_kw:        if condition in text_lower:            conditions.append(condition.capitalize())        for med in med_kw:        if med in text_lower:            medications.append(med.capitalize())        summary = f"""**Chief Complaint:** {', '.join(symptoms) if symptoms else 'As described'}**Present Illness:**{text[:200]}{'...' if len(text) > 200 else ''}**Known Conditions:** {', '.join(conditions) if conditions else 'None documented'}**Current Medications:** {', '.join(medications) if medications else 'None documented'}**Clinical Assessment:** Patient presents with symptoms requiring evaluation for treatment options and potential clinical trial enrollment.*Note: AI-generated summary for demonstration purposes. Clinical validation required.*"""        return summarydef display_trial_matches(matches: List[Dict], limit: int = 3):    """Display matched clinical trials"""    if not matches:        print("⚠️  No matching trials found")        return        print(f"\n🔬 Found {len(matches)} Matching Clinical Trials")    print("=" * 80)        for i, trial in enumerate(matches[:limit], 1):        print(f"\n[{i}] {trial['title']}")        print(f"    Trial ID: {trial['trial_id']}")        print(f"    Condition: {trial['condition']}")        print(f"    Status: {trial['status']}")                if 'similarity_score' in trial:            score_percent = trial['similarity_score'] * 100            bar_length = int(score_percent / 5)            bar = "█" * bar_length + "░" * (20 - bar_length)            print(f"    Match Score: {score_percent:.1f}% {bar}")                print(f"    Phase: {trial.get('phase', 'N/A')}")        print(f"    Enrollment: {trial.get('enrollment', 'N/A')}")        print(f"    Description: {trial['description'][:120]}...")        print(f"    Eligibility: {trial['eligibility'][:100]}...")        print(f"    Location: {trial['location']}")        print("\n" + "=" * 80)def measure_performance(func, *args, **kwargs):    """Measure function performance"""    process = psutil.Process(os.getpid())    start_memory = process.memory_info().rss / 1024 / 1024    start_time = time.time()        result = func(*args, **kwargs)        end_time = time.time()    end_memory = process.memory_info().rss / 1024 / 1024        metrics = {        "execution_time_ms": round((end_time - start_time) * 1000, 2),        "memory_start_mb": round(start_memory, 2),        "memory_end_mb": round(end_memory, 2),        "memory_delta_mb": round(end_memory - start_memory, 2)    }        return result, metricsprint("✅ Helper functions defined:")print("   • strip_pii() - PII detection and removal")print("   • generate_demo_summary() - Demo medical summarization")print("   • display_trial_matches() - Format trial results with progress bars")print("   • measure_performance() - Performance and memory metrics")
```
